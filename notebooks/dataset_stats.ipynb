```python
import pandas as pd

df_titanic = pd.read_csv("../data/titanic.csv")
df_estonia = pd.read_csv("../data/estonia.csv")
df_lusitania = pd.read_csv("../data/lusitania.csv")

df_all_ships = pd.read_csv("../data/all_ships.csv")

print(f"Dimensione dataset titanic: {df_titanic.shape}")
print(f"Dimensione dataset estonia: {df_estonia.shape}")    
print(f"Dimensione dataset lusitania: {df_lusitania.shape}")
print(f"Dimensione dataset all_ships: {df_all_ships.shape}")
```

```python
print(f"Nomi variabili dataset titanic: {df_titanic.columns}")
print(f"Nomi variabili dataset estonia: {df_estonia.columns}")
print(f"Nomi variabili dataset lusitania: {df_lusitania.columns}")
print(f"Nomi variabili dataset all_ships: {df_all_ships.columns}")
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

survived_counts = df_all_ships['survived'].value_counts()
total = survived_counts.sum()

plt.figure(figsize=(6, 4))
plt.bar(survived_counts.index, survived_counts.to_numpy(), color=['red', 'green'])

plt.title('Survival distribution')
plt.xticks([0, 1], ['Not survived', 'Survived'])
plt.ylabel('Number of passengers')

plt.savefig('../media/survival_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print("Stats for report")
print(f"Survivors count: {survived_counts[1]}")
print(f"Dead count: {survived_counts[0]}")
print(f"Survival chance: {survived_counts[1] / total :.4f}")
print(f"Death chance: {survived_counts[0] / total :.4f}")
```

```python
sex_counts = df_all_ships['sex'].value_counts()
total = sex_counts.sum()

plt.figure(figsize=(6, 4))
plt.bar(sex_counts.index, sex_counts.to_numpy(), color=['blue', 'pink'])

plt.title('Sex distribution')
plt.xticks([0, 1], ['Male', 'Female'])
plt.ylabel('Number of passengers')

plt.savefig('../media/sex_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print("Stats for report")
print(f"Male count: {sex_counts[0]}")
print(f"Female count: {sex_counts[1]}")
print(f"Male proportion: {sex_counts[0] / total :.4f}")
print(f"Female proportion: {sex_counts[1] / total :.4f}")
```

```python
male_counts = df_all_ships[df_all_ships['sex'] == 0]['survived'].value_counts()
female_counts = df_all_ships[df_all_ships['sex'] == 1]['survived'].value_counts()
total_male = male_counts.sum()
total_female = female_counts.sum()

# Il grafico presenta barre impilate
plt.figure(figsize=(8, 4))
plt.bar(0, male_counts[0], color='red')
plt.bar(0, male_counts[1], bottom=male_counts[0], color='green')
plt.bar(1, female_counts[0], color='red')
plt.bar(1, female_counts[1], bottom=female_counts[0], color='green')

plt.title('Survival distribution by sex')
plt.xticks([0, 1], ['Male', 'Female'])
plt.ylabel('Number of passengers')

legend_elements = [Patch(facecolor='green', label='Survived'), Patch(facecolor='red', label='Not survived')]
plt.legend(handles=legend_elements)

plt.savefig('../media/survival_by_sex.png', dpi=300, bbox_inches='tight')
plt.show()

print("Stats for report")
print(f"Male alive count: {male_counts[1]}")
print(f"Male dead count: {male_counts[0]}")
print(f"Female alive count: {female_counts[1]}")
print(f"Female dead count: {female_counts[0]}\n")

print(f"Male alive chance: {male_counts[1] / total_male :.4f}")
print(f"Male dead chance: {male_counts[0] / total_male :.4f}")
print(f"Female alive chance: {female_counts[1] / total_female :.4f}")
print(f"Female dead chance: {female_counts[0] / total_female :.4f}")
```

```python
male_counts = df_titanic[df_titanic['Sex'] == 'male']['Survived'].value_counts()
female_counts = df_titanic[df_titanic['Sex'] == 'female']['Survived'].value_counts()
legend_elements = [Patch(facecolor='green', label='Survived'), Patch(facecolor='red', label='Not survived')]

plt.figure(figsize=(6, 3))
plt.bar(0, male_counts[0], color='red')
plt.bar(0, male_counts[1], bottom=male_counts[0], color='green')
plt.bar(1, female_counts[0], color='red')
plt.bar(1, female_counts[1], bottom=female_counts[0], color='green')

plt.title('Survival distribution by sex - Titanic')
plt.xticks([0, 1], ['Male', 'Female'])
plt.ylabel('Number of passengers')

plt.legend(handles=legend_elements)

plt.savefig('../media/survival_by_sex_titanic.png', dpi=300, bbox_inches='tight')
plt.show()

male_counts = df_estonia[df_estonia['Sex'] == 'M']['Survived'].value_counts()
female_counts = df_estonia[df_estonia['Sex'] == 'F']['Survived'].value_counts()

plt.figure(figsize=(6, 3))
plt.bar(0, male_counts[0], color='red')
plt.bar(0, male_counts[1], bottom=male_counts[0], color='green')
plt.bar(1, female_counts[0], color='red')
plt.bar(1, female_counts[1], bottom=female_counts[0], color='green')

plt.title('Survival distribution by sex - Estonia')
plt.xticks([0, 1], ['Male', 'Female'])
plt.ylabel('Number of passengers')

plt.legend(loc='lower left', handles=legend_elements)

plt.savefig('../media/survival_by_sex_estonia.png', dpi=300, bbox_inches='tight')
plt.show()

male_counts = df_lusitania[df_lusitania['Sex'] == 'Male']['Fate'].value_counts()
female_counts = df_lusitania[df_lusitania['Sex'] == 'Female']['Fate'].value_counts()

plt.figure(figsize=(6, 3))
plt.bar(0, male_counts['Lost'], color='red')
plt.bar(0, male_counts['Saved'], bottom=male_counts['Lost'], color='green')
plt.bar(1, female_counts['Lost'], color='red')
plt.bar(1, female_counts['Saved'], bottom=female_counts['Lost'], color='green')

plt.title('Survival distribution by sex - Lusitania')
plt.xticks([0, 1], ['Male', 'Female'])
plt.ylabel('Number of passengers')

plt.legend(handles=legend_elements)

plt.savefig('../media/survival_by_sex_lusitania.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
age = df_all_ships[df_all_ships['age_missing'] == 0]['age'] # si considerano solo le età realmente presenti (age_missing = 0)

plt.figure(figsize=(8, 4))
plt.hist(age, bins=80, color='skyblue', edgecolor='black')

plt.title('Age distribution of passengers')
plt.xticks(range(0, 88, 5))
plt.xlabel('Age')
plt.ylabel('Number of passengers')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.savefig('../media/age_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

age_survived = df_all_ships[(df_all_ships['age_missing'] == 0) & (df_all_ships['survived'] == 1)]['age']
age_not_survived = df_all_ships[(df_all_ships['age_missing'] == 0) & (df_all_ships['survived'] == 0)]['age']

plt.figure(figsize=(8,4))
plt.hist([age_not_survived, age_survived], bins=80, color=['red', 'green'], label=['Not survived', 'Survived'], stacked=True)

plt.title('Age distribution by survival')
plt.xticks(range(0, 88, 5))
plt.xlabel('Age')
plt.ylabel('Number of passengers')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.savefig('../media/age_distribution_by_survival.png', dpi=300, bbox_inches='tight')
plt.show()
```
